# Tapşırıq 1 - Kitablar

```python
books = pd.DataFrame({
    'Title': ['1984', 'Sapiens', 'Atomic Habits', 'Clean Code', 'Dune', 'The Alchemist'],
    'Author': ['George Orwell', 'Yuval Harari', 'James Clear', 'Robert Martin', 'Frank Herbert', 'Paulo Coelho'],
    'PublicationYear': [1949, 2011, 2018, 2008, 1965, 1988],
    'Price': [12.5, 18.0, 20.0, 35.0, 15.0, 10.0]
})
books
```

```python
print('Orta qiymət:', books['Price'].mean())
```

```python
books[books['PublicationYear'] > 2015]
```

```python
books.sort_values(by='Price')
```

# Tapşırıq 3 - Qida məhsulları

```python
food = pd.DataFrame({
    'Product': ['Apple', 'Banana', 'Chicken', 'Rice', 'Egg', 'Cheese', 'Bread', 'Salmon', 'Nuts', 'Chocolate'],
    'Category': ['Fruit', 'Fruit', 'Meat', 'Grain', 'Dairy', 'Dairy', 'Grain', 'Fish', 'Snack', 'Snack'],
    'Calories': [52, 89, 239, 130, 155, 402, 265, 208, 607, 546],
    'Protein': [0.3, 1.1, 27.0, 2.7, 13.0, 25.0, 9.0, 20.0, 20.0, 4.9]
})
food
```

```python
food[food['Calories'] > 300]
```

```python
food.groupby('Category')['Protein'].mean()
```

```python
food.sort_values(by='Calories', ascending=False)
```

# Tapşırıq 4 - Əməkdaşlar və layihələr

```python
work = pd.DataFrame({
    'Name': ['Ali', 'Leyla', 'Murad', 'Ali', 'Leyla', 'Nigar', 'Murad', 'Nigar'],
    'Project': ['Website', 'Mobile App', 'Website', 'CRM', 'Website', 'CRM', 'Mobile App', 'Website'],
    'Hours': [20, 35, 15, 10, 12, 25, 30, 8]
})
work
```

```python
work.groupby('Name')['Hours'].sum()
```

```python
work.groupby('Project')['Hours'].sum()
```

```python
hours_by_name = work.groupby('Name')['Hours'].sum()
print('Ən çox işləyən:', hours_by_name.idxmax(), '-', hours_by_name.max(), 'saat')
```

# Tapşırıq 5 - Bilet satışları

```python
tickets = pd.DataFrame({
    'Movie': ['Inception', 'Avatar', 'Titanic', 'Inception', 'Avatar', 'Titanic',
              'Inception', 'Avatar', 'Titanic', 'Inception', 'Avatar', 'Titanic'],
    'City': ['Baku', 'Baku', 'Baku', 'Ganja', 'Ganja', 'Ganja',
             'Sumgait', 'Sumgait', 'Sumgait', 'Baku', 'Ganja', 'Sumgait'],
    'TicketsSold': [120, 150, 90, 60, 80, 40, 50, 70, 30, 100, 65, 25]
})
tickets
```

```python
tickets.groupby('Movie')['TicketsSold'].sum()
```

```python
tickets.groupby('City')['TicketsSold'].sum()
```

```python
by_movie = tickets.groupby('Movie')['TicketsSold'].sum()
print('Ən çox bilet satılan film:', by_movie.idxmax(), '-', by_movie.max())
```
