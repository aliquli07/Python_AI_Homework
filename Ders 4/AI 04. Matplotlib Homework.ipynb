# AI 04 - Matplotlib ev tapşırıqları

```python
import numpy as np
import matplotlib.pyplot as plt
```

# Tapşırıq 1 - Həftəlik temperatur qrafiki

```python
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
temperature = [18, 20, 22, 21, 24, 26, 23]
```

```python
plt.figure(figsize=(10, 5))
plt.plot(days, temperature, color='orange', marker='o', linestyle='--', linewidth=2, label='Temperature')

# her noqtenin ustunde deyeri yaziram
for i in range(len(days)):
    plt.text(days[i], temperature[i] + 0.3, f'{temperature[i]}°C', ha='center')

plt.title('Weekly Temperature')
plt.xlabel('Days')
plt.ylabel('Temperature (°C)')
plt.grid(True)
plt.legend()
plt.show()
```

# Tapşırıq 2 - Proqramlaşdırma dillərinin müqayisəsi

```python
languages = ["Python", "JavaScript", "Java", "C#", "C++"]
students = [35, 28, 22, 18, 15]
```

```python
colors = ['gold', 'skyblue', 'lightgreen', 'violet', 'salmon']

max_index = students.index(max(students))

plt.figure(figsize=(10, 5))
bars = plt.barh(languages, students, color=colors)

# en cox telebesi olan dili vurgulayiram
bars[max_index].set_edgecolor('red')
bars[max_index].set_linewidth(3)

for i in range(len(languages)):
    plt.text(students[i] + 0.5, i, str(students[i]), va='center')

plt.title('Students by Programming Language')
plt.xlabel('Number of students')
plt.ylabel('Languages')
plt.gca().invert_yaxis()
plt.show()
```

## Əlavə - vertical bar chart

```python
plt.figure(figsize=(8, 5))
bars = plt.bar(languages, students, color=colors)
bars[max_index].set_edgecolor('red')
bars[max_index].set_linewidth(3)

for i in range(len(languages)):
    plt.text(i, students[i] + 0.5, str(students[i]), ha='center')

plt.title('Students by Programming Language')
plt.xlabel('Languages')
plt.ylabel('Number of students')
plt.show()
```

# Tapşırıq 3 - Tələbələrin nəticələrinin analizi

```python
study_hours = [1, 2, 2, 3, 4, 4, 5, 6, 7, 8]
scores = [45, 50, 55, 60, 65, 70, 74, 80, 88, 95]
ages = [18, 19, 18, 20, 21, 19, 22, 20, 23, 21]
```

```python
fig = plt.figure(figsize=(10, 6))
ax = fig.add_axes([0.1, 0.1, 0.8, 0.8])

sc = ax.scatter(study_hours, scores, c=ages, cmap='viridis', marker='o', s=120)
cbar = fig.colorbar(sc, ax=ax)
cbar.set_label('Age')

# en yuksek netice
best = scores.index(max(scores))
ax.scatter(study_hours[best], scores[best], color='red', marker='*', s=400, label='Best score')
ax.annotate(f'Max: {scores[best]}', (study_hours[best], scores[best]),
            xytext=(study_hours[best] - 2, scores[best] - 3),
            arrowprops=dict(arrowstyle='->'))

ax.set_title('Study Hours vs Exam Score')
ax.set_xlabel('Study Hours')
ax.set_ylabel('Exam Score')
ax.grid(True)
ax.legend(loc='lower right')

# elave - kicik axes
ax2 = fig.add_axes([0.15, 0.55, 0.25, 0.25])
ax2.bar(range(1, len(scores) + 1), scores, color='gray')
ax2.set_title('Scores', fontsize=9)
ax2.tick_params(labelsize=7)

plt.show()
```
